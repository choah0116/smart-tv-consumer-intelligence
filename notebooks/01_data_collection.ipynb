# Smart TV Data Collection

This notebook collects product metadata and consumer reviews for five major Smart TV brands — Samsung, LG, Sony, TCL, and Hisense — from the Amazon Reviews 2023 dataset.

The collection process identifies television products using product category metadata, standardizes brand names, and extracts consumer reviews associated with the selected products through `parent_asin`.

```python
import pandas as pd
from datasets import load_dataset

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 200)
```

## Product Metadata Collection

Product metadata is collected from the Electronics category of the Amazon Reviews 2023 dataset. Each metadata shard is processed separately to reduce memory usage.

Products are retained when their category metadata includes `Televisions`, and the sample is restricted to Samsung, LG, Sony, TCL, and Hisense. Brand names are standardized during collection for consistent downstream analysis.

```python
target_stores = {
    "SAMSUNG": "Samsung",
    "LG": "LG",
    "Sony": "Sony",
    "TCL": "TCL",
    "Hisense": "Hisense"
}

all_tv_products = []

for i in range(10):
    print(f"Processing metadata shard {i + 1}/10...")

    url = (
        "https://huggingface.co/datasets/"
        "McAuley-Lab/Amazon-Reviews-2023/resolve/main/"
        f"raw_meta_Electronics/full-{i:05d}-of-00010.parquet"
    )

    # Load one shard at a time
    shard = load_dataset(
        "parquet",
        data_files=url,
        split="train"
    )

    # Keep only actual televisions
    shard = shard.filter(
        lambda row: (
            row["categories"] is not None
            and "Televisions" in row["categories"]
        )
    )

    # Keep only our five target brands
    shard = shard.filter(
        lambda row: row["store"] in target_stores
    )

    # Convert only the filtered rows to pandas
    shard_df = shard.to_pandas()

    # Standardize brand names
    shard_df["brand"] = shard_df["store"].map(target_stores)

    all_tv_products.append(shard_df)

    print(f"  Found {len(shard_df):,} target TVs")
```

### Combine and Validate Product Metadata

The filtered metadata shards are combined into a single product-level dataset. Basic validation checks are used to confirm the number of products, unique product identifiers, and brand distribution.

```python
tv_products = pd.concat(
    all_tv_products,
    ignore_index=True
)

print("Total rows:", len(tv_products))
print("Unique parent_asin:", tv_products["parent_asin"].nunique())

tv_products["brand"].value_counts()
```

```python
duplicates = tv_products["parent_asin"].duplicated().sum()

print("Duplicate parent_asin:", duplicates)
```

```python
brand_summary = (
    tv_products
    .groupby("brand")
    .agg(
        products=("parent_asin", "nunique"),
        total_ratings=("rating_number", "sum"),
        avg_product_rating=("average_rating", "mean")
    )
    .sort_values("products", ascending=False)
)

brand_summary
```

### Save Product Dataset

The validated product metadata is saved as a Parquet file for use in the subsequent cleaning and analysis notebooks.

```python
tv_products.to_parquet(
    "../data/tv_products.parquet",
    index=False
)

print("Saved ../data/tv_products.parquet")
```

## Consumer Review Collection

Consumer reviews are collected for the Smart TV products identified above. Reviews are matched to products using `parent_asin`.

Because the full Electronics review dataset is large, review shards are downloaded and filtered individually. Intermediate checkpoints are saved so that completed shards do not need to be processed again.

```python
import os
import requests
import pyarrow as pa
import pyarrow.parquet as pq
import pyarrow.compute as pc

os.makedirs("review_checkpoints", exist_ok=True)
os.makedirs("temp_reviews", exist_ok=True)

COMMIT = "ac9d3ad3342d6f00bf6ad8caa2668a3f830e2dee"

review_columns = [
    "rating",
    "title",
    "text",
    "asin",
    "parent_asin",
    "user_id",
    "timestamp",
    "helpful_vote",
    "verified_purchase"
]

tv_asin_array = pa.array(list(tv_asins))
```

```python
def download_file(url, output_path):
    with requests.get(url, stream=True) as r:
        r.raise_for_status()

        with open(output_path, "wb") as f:
            for chunk in r.iter_content(chunk_size=1024 * 1024):
                if chunk:
                    f.write(chunk)
```

```python
for i in range(34):

    output_file = f"review_checkpoints/tv_reviews_{i:02d}.parquet"
    temp_file = f"temp_reviews/review_{i:02d}.parquet"

    # Already completed → skip
    if os.path.exists(output_file):
        print(f"Shard {i + 1}/34 already completed — skipping")
        continue

    print(f"\nProcessing review shard {i + 1}/34...")

    url = (
        "https://huggingface.co/datasets/"
        "McAuley-Lab/Amazon-Reviews-2023/resolve/"
        f"{COMMIT}/"
        f"raw_review_Electronics/full-{i:05d}-of-00034.parquet"
    )

    # 1. Download shard temporarily
    print("  Downloading...")
    download_file(url, temp_file)

    # 2. Open parquet without loading entire shard into RAM
    parquet_file = pq.ParquetFile(temp_file)

    matched_batches = []
    matched_count = 0

    # 3. Process in batches
    for batch in parquet_file.iter_batches(
        batch_size=100_000,
        columns=review_columns
    ):

        table = pa.Table.from_batches([batch])

        mask = pc.is_in(
            table["parent_asin"],
            value_set=tv_asin_array
        )

        matched = table.filter(mask)

        if matched.num_rows > 0:
            matched_batches.append(matched)
            matched_count += matched.num_rows

    # 4. Combine matched TV reviews for this shard
    if matched_batches:
        shard_reviews = pa.concat_tables(matched_batches)

    else:
        # Create empty table with correct schema
        schema = parquet_file.schema_arrow

        selected_schema = pa.schema(
            [schema.field(col) for col in review_columns]
        )

        shard_reviews = pa.Table.from_batches(
            [],
            schema=selected_schema
        )

    # 5. Save checkpoint
    pq.write_table(
        shard_reviews,
        output_file
    )

    print(f"  Saved {matched_count:,} TV reviews")

    # 6. Delete huge original shard from Colab disk
    del parquet_file

    if os.path.exists(temp_file):
        os.remove(temp_file)

    print("  Temporary shard deleted")
```

### Combine and Validate Consumer Reviews

The filtered review checkpoints are combined into a single dataset. Reviews are mapped to their corresponding brands using `parent_asin`, and validation checks are performed to identify missing brand assignments and summarize review coverage.

```python
import glob

checkpoint_files = sorted(
    glob.glob("review_checkpoints/tv_reviews_*.parquet")
)

print("Checkpoint files:", len(checkpoint_files))
```

```python
tv_reviews = pd.concat(
    [pd.read_parquet(f) for f in checkpoint_files],
    ignore_index=True
)

print("Total TV reviews:", len(tv_reviews))
print("Unique TV products:", tv_reviews["parent_asin"].nunique())
```

```python
brand_map = (
    tv_products
    .set_index("parent_asin")["brand"]
    .to_dict()
)

tv_reviews["brand"] = tv_reviews["parent_asin"].map(brand_map)

tv_reviews["brand"].value_counts()
```

```python
print("Missing brands:", tv_reviews["brand"].isna().sum())
```

```python
review_summary = (
    tv_reviews
    .groupby("brand")
    .agg(
        reviews=("rating", "size"),
        avg_rating=("rating", "mean"),
        unique_products=("parent_asin", "nunique"),
        verified_purchase_rate=("verified_purchase", "mean"),
        avg_helpful_votes=("helpful_vote", "mean")
    )
    .sort_values("reviews", ascending=False)
)

review_summary
```
